```python
# Grade Calculator
```

```python
marks = int(input("Enter marks: "))
```

```python
if marks >= 90:
        grade = "A+"
elif marks >= 80:
        grade = "A"
elif marks >= 70:
        grade = "B"
elif marks >= 60:
        grade = "C"
elif marks >= 50:
        grade = "D"  # Pass
else:
        grade = "F"  # Fail
```

```python
print("Grade:", grade)
```
