```python
balance = 500000 #PKR
daily_limit = 25000
pin = "1234"
```

```python
entered_pin = input("Enter PIN: ")
```

```python
if entered_pin == pin:
		amount = int(input("Withdraw amount: "))
		if amount <= 0:
			print("Invalid amount!")
		elif amount > daily_limit:
			print(f"Exceeds daily limit of {daily_limit}")
		elif amount > balance:
			print("insufficient balance!")
		else:
			  balance-=amount
			  print(f"withdraw{amount}.Balance: {balance}")
else:
		print("Wrong PIN! Card Blocked.")
```
