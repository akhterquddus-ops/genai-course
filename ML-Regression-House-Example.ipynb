```python
# 1. Make up some fake data (No CSV needed!)
import pandas as pd
data = {'area': [1000, 1500, 2000, 2500, 3000],
        'price': [200000, 300000, 400000, 500000, 600000]}
df = pd.DataFrame(data)

X = df[['area']] # The Clue
y = df['price']  # The Answer

# 2. Split it
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 3. Train it
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train) # The "Studying" phase

# 4. Predict and Grade
predictions = model.predict(X_test)
print("Model guessed:", predictions)
print("Real price was:", y_test.values)
```
