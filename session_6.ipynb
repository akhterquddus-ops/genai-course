```python
 def get_grade(marks):

    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"
```

```python
print(get_grade(85))  # A
```

```python
import random, string
```

```python
def gen_password(length=12):
    chars = string.ascii_letters
    chars += string.digits + "!@#$"
    pwd = "".join(random.choices(
        chars, k=length))
    return pwd
```

```python
 print(gen_password)
```
