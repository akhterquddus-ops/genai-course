```python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.marks = []

    def add_mark(self, mark):
        if 0 <= mark <= 100:
            self.marks.append(mark)
        else:
            print("Mark must be between 0 and 100")

    def average(self):
        return sum(self.marks) / len(self.marks) if self.marks else 0

    def get_grade(self):
        avg = self.average()

        if avg >= 90:
            return "A"
        elif avg >= 80:
            return "B"
        elif avg >= 70:
            return "C"
        elif avg >= 60:
            return "D"
        else:
            return "F"

    def __str__(self):
        return f"{self.name} ({self.student_id}) — Grade: {self.get_grade()}"


# Create student
student1 = Student("Ali", "S001")

# Add marks
student1.add_mark(85)
student1.add_mark(90)
student1.add_mark(78)

# Display information
print("Name:", student1.name)
print("Student ID:", student1.student_id)
print("Marks:", student1.marks)
print("Average:", student1.average())
print("Grade:", student1.get_grade())

# Print the object
print(student1)
```
